```python
raw_ticket_price = "150.50"
raw_guests_count = "45"
raw_event_id = "FEST-2026"
#Convert the raw data to appropriate types
ticket_price = float(raw_ticket_price)
guests_count = int(raw_guests_count)
total_revenue = ticket_price * guests_count
print(f"Event ID: {raw_event_id}")
print(f"Total Revenue: ${total_revenue:.2f}")
```
