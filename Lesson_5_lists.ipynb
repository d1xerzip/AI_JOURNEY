```python
# Copy (copy by reference) the list A
A = ["hard rock", 10, 1.2] # A seted
B = A #Create another icon
print('A:', A) # a result
print('B:', B) # b same result
print('B[0]:', B[0]) # Becouse b = a, 0 = Hard rock = B[0]
A[0] = "BANANA" #change value A[0]
print('B[0]:', B[0]) #Becouse b = a its changing in b ass well
B = A[:] #making a clone
B # ['BANANA', 10, 1.2]
print('B[0]:', B[0])
A0 = "HARD ROCK"
print('B[0];', B[0])
```
