```python
album_set1 = set(["Thriller", 'AC/DC', 'Back in Black'])
album_set2 = set (["AC/DC", "Back in Black", "The Dark Side of the Moon"])
album_set1, album_set2
intersection = album_set1 & album_set2 #same from 1 and 2
intersection 
album_set1.difference(album_set2)# dif between 1 and 2
album_set2.difference(album_set1)# dif between 2 an 1
album_set1.intersection(album_set2) #same from 1 and 2
album_set1.union(album_set2) # all 1 and 2
set(album_set1).issubset(album_set2) # are all from set 1 in set 2? (false)
set(album_set2).issubset(album_set1)   # are all from set 2 in set 1? (false)  
set(intersection).issubset(album_set1) # is intersection all in set 1 (true)
set(intersection).issubset(album_set2) # is intersection all in set 2 (true)
set(intersection).issuperset(album_set1) 
set(intersection).issuperset(album_set2)
set(album_set1).issuperset(album_set2)
set(album_set2).issuperset(album_set1)
set({"Back in Black", "AC/DC"}).issubset(album_set1) 
album_set1.issuperset({"Back in Black", "AC/DC"})

  
```
